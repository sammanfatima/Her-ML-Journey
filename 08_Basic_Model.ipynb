# Model Notebook in Sklearn

```python
import pandas as pd
import numpy as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LinearRegression
```

```python
# import the data
df = sns.load_dataset('tips')
df.head()
```

```python
# separate the features(X) and the target/labels(y)

X = df[['total_bill']] # X always 2D
y = df['tip']
```

```python
# train test split data
X_train, X_test, y_train, y_test = train_test_split(X, y ,train_size=.8)
```

* fit calculates the mean and standard deviation of the training data.

**transform**
* It uses those values to change/scale the data so that the features are on a similar scale.
* it uses the same mean and standard deviation learned from X_train

* Fit → calculates/learns the mean and standard deviation from X_train.

* Transform → uses those mean and standard deviation values to scale X_train

```python
# preprocess the data
scalar = StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)

```

* when building an ML model, we fit the model on the **training data**
* The model learns patterns from the training data.

```python
# build the model
model = LinearRegression()
model.fit(X_train, y_train)
```

* We give the test data to our train model and model gives us its prediction

```python
# prediction
y_pred = model.predict(X_test)
```

* help us check how well our model is predicting.
* **MSE** How much wrong our predictions are
* If your predicted value is far away from the actual value, the error will be larger.

* **R2-score** How well our model explains/predicts the actual values.

```python
# evaluate the model
from sklearn.metrics import mean_squared_error, r2_score
print('MSE:', mean_squared_error(y_test, y_pred))
print('R2:', r2_score(y_test, y_pred))
```

# Save and Load model

```python
# Save the model
import pickle
pickle.dump(model, open('model.pkl', 'wb'))
```
