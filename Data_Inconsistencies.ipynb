# Data Inconsistencies

1. Incocnistent format
* for example
* DD-MM-YYY, YYYY-MM-DD, DD/MM/YYYY
* Find out which format is important for you

2. Inconsistent Naming Conventions
* Data collected for countries sometimes it is written as USA, United States of America... you have to make these in one consistent form

3. Typographical errors
* Mistakes in data entries

4. Duplication
* one rows show again

5. contradictory data


```python
import pandas as pd
```

```python
df = {
    'date': ['2021-12-01', '01-12-2022', '2022/12/01', '12-01-2021'],
    'country': ['USA', 'U.S.A.', 'America', 'United States'],
    'name': ['John Doe', 'Jonh Doe', 'Jane Doe', 'Jane Doe'],
    'sales_2020': [100, 200, None, 200],
    'sales_2021': [None, 150, 300, 150]
}
# make pandas dataframes
df = pd.DataFrame(df)
```

```python
df.head()
```

# Standardizing date format

* Convert the date col into a proper date format
* df['date'] → selects the date column
* pd.to_datetime() → converts values into date format
* errors='coerce' → If a value is not a valid date, replace it with a missing value instead of showing an error.

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df.head()
```

* It changes the date into specific format: Year-Month-Day

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['date'] = df['date'].dt.strftime('%Y-%m-%d')
df.head()
```

# Replace a specific value

```python
df.head()
```

```python
df.loc[3, 'date'] = '2022-12-08'
```

```python
df.head()
```

# Replace all missing date with same date

```python
df['date'] = df['date'].fillna('2002-03-08')
```

```python
df.head()
```

# Make all country names follow the same format

```python
country_mapping ={'USA': 'United States','U.S.A.': 'United States', 'America': 'United States'}
df['country']= df['country'].replace(country_mapping)
df.head()
```

# Correct the typographical mistakes in name

```python
df['name'] = df['name'].replace({'Jonh Doe': 'John Doe'})
df.head()
```

```python
df.loc[2, 'name'] ='Sara'
df.head()
```

# Remove duplicates

```python
df = df.drop_duplicates(subset="name")
df.head()
```
