# Feature Encoding
* Feature encoding is the process of converting categorical (text) data into numerical data so that a machine learning model can understand it.

# Popular method of Encoding
* Label encoding
* Ordinal encoding
* One hot encoding
* Binary encoding

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load data
df = sns.load_dataset('tips')
df.head()
```

# Value count of time

```python
df['time'].value_counts()
```

* Dinner → 176 = Dinner appears 176 times in the time column.
* Lunch → 68 = Lunch appears 68 times in the time column.

# Value count of day

```python
df['day'].value_counts()
```

# Value count of smoker

```python
df['smoker'].value_counts()
```

# Value count of sex

```python
df['smoker'].value_counts()
```

# Encode the time in labelencoder with sklearn

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['encoded_time']= le.fit_transform(df['time'])
df.head()
```

```python
df['encoded_time'].value_counts() # dinner= 0, lunch=1
```

# Ordinal encoding the day column using specific order

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=[['Thur', 'Fri', 'Sat', 'Sun']])

df['encoded_day'] = oe.fit_transform(df[['day']])

df.head()
```

```python
df['encoded_day'].value_counts() # 1.0= thurs, 2.0= fri, 3.0= Sat, 4.0= Sun
```

```python
oe = OrdinalEncoder()

df['encoded_day'] = oe.fit_transform(df[['day']])

df.head()
```

```python
df['encoded_day'].value_counts() # encode randomly
```

# One hot encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
ohe.fit_transform(df[['smoker']]).toarray()
```
