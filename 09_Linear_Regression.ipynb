# Linear Regression
* Linear Regression is a supervised machine learning algorithm that uses a straight line to predict a numerical value.
* **Simple regression** only one independent variable(e.g, sales)
* **Multiple regression** more than one indenpendent varibale(e.g, sales, price)

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

```

```python
# load the dataset
df = sns.load_dataset('tips')
df.head()
```

```python
sns.scatterplot(x='total_bill', y='tip', data=df)
```

```python
# split the data into X and y
X = df[['total_bill']]
scalar = MinMaxScaler()
X = scalar.fit_transform(X)
y = df['tip']
```

```python
# split the data into train and test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
# call the model
model = LinearRegression()
```

```python
# train the model
model.fit(X_train, y_train)
```

```python
# take out model intercept and slope make an equation
print(model.intercept_) # starting value
print(model.coef_) # slope -> how much it changes
print('y =', model.intercept_, '+', model.coef_, '*X')
```

```python
# predict
y_pred = model.predict(X_test)
```

```python
# evaluate the model
print('MSE:', mean_squared_error(y_test, y_pred))
print('R2:', r2_score(y_test, y_pred))
print('RMSE:', np.sqrt(mean_squared_error(y_test, y_pred)))
```

```python
# plot the model
plt.scatter(X_test, y_test)
plt.plot(X_test, y_pred, color='green')
plt.show()
```
