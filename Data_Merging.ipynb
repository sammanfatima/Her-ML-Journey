# Data Merging

```python
import pandas as pd
```

# Dataset 1:

```python
data1 ={
    'id' : [1, 2, 3],
    'name': ['Ahmed', 'Fatima', 'Ali'],
    'age': [28, 32, 24]
}
df1 = pd.DataFrame(data1)
df1.head()
```

# Dataset 2:





```python
data2 = {
    'id': [1, 2, 3],
    'city':['Karachi', 'Lahore', 'Islamabad'],
    'occupation': ['Engineer', 'Teacher', 'Developer']
}
df2 = pd.DataFrame(data2)
df2.head()
```

# Merging dataframes on 'id'

* pd.merge() is used to combine two dataframes together based on a common column

```python
merge_df = pd.merge(df1, df2, on='id', how='inner')
merge_df.head()
```
